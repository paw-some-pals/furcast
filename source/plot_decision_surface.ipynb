```python
import pickle
import pandas as pd
from sklearn.tree import plot_tree
import matplotlib.pyplot as plt
from matplotlib.colors import ListedColormap
from sklearn.inspection import DecisionBoundaryDisplay
from sklearn.tree import DecisionTreeClassifier
```

```python
# Open the file in binary read mode ('rb')
with open('../models/model_dog_aac.pkl', 'rb') as file:
    loaded_model = pickle.load(file)

# Load data
df = pd.read_csv("../datasets/dog_train_aac.csv")
target_col = df.pop('stay_category') 
```

```python
# Use the model to make predictions
# 2. Set up the figure size (Random Forest trees can be very large)
plt.figure(dpi=1000)

# 3. Plot the hierarchical diagram
plot_tree(
    loaded_model[-1].estimators_[0],
    feature_names=df.columns,
    class_names=target_col.unique().tolist(),
    filled=True,
    fontsize=5,
    rounded=True,
    max_depth=2
)

plt.title("Diagram of Single Tree From Random Forest", fontsize=12)
plt.tight_layout()
plt.savefig('../figures/one_tree.png', transparent=True, bbox_inches='tight')
plt.show()
```
